## 1. Введение

### 1.1. Актуальность

Социальное доверие - один из ключевых нематериальных ресурсов общества. Оно связано с устойчивостью демократических институтов, эффективностью коллективных действий и экономическим развитием (Putnam 2000; Rothstein and Uslaner 2005). Распределение доверия между странами крайне неравномерно: в странах Северной и Западной Европы уровень генерализованного доверия существенно выше, чем в странах Юга и Востока. Причины этого разрыва обсуждаются десятилетиями, и однозначного ответа пока нет.

Одним из основных индивидуальных коррелятов доверия выступают базовые ценности. Теория Шалома Шварца (Schwartz 1992) описывает универсальную структуру из десяти первичных ценностей, группирующихся в четыре высших измерения. Проблема в том, что её эмпирическая применимость к сопоставлению стран требует проверки измерительной инвариантности: без неё нельзя корректно сравнивать ценности между государствами.



### 1.2. Проблема исследования

В большинстве работ по доверию и ценностям в Европе либо берут одну страну (теряется кросс-национальный контекст), либо запускают обычные регрессии без учёта вложенности индивидов в страны. Измерительная модель Шварца на объединённой выборке редко проверяется явно. Эти два пробела приводят к риску смешения индивидуальных и страновых эффектов и к неустойчивой интерпретации коэффициентов.

Мы решаем задачу в два шага: сначала проверяем структуру ценностей Шварца через конфирматорный факторный анализ (CFA, confirmatory factor analysis) и тест инвариантности между странами, затем используем валидированные индексы в многоуровневой модели с индивидуальными и страновыми предикторами. Дополнительно ставим рядом модель машинного обучения (ML, machine learning) той же зависимой переменной, чтобы посмотреть, сходятся ли выводы двух методологически разных подходов.



### 1.3. Объект и предмет исследования

**Объект:** население 29 стран Европы, представленное в данных European Social Survey Round 9 (2018).

**Предмет:** структура базовых ценностей по Шварцу и её связь с уровнем генерализованного социального доверия на индивидуальном и страновом уровнях.



### 1.4. Цель и задачи

**Цель:** установить, какие индивидуальные и страновые факторы связаны с уровнем социального доверия в Европе, с учётом иерархической структуры данных и предварительной валидации измерительной модели ценностей.

**Задачи:**
1. Отобрать и очистить данные ESS Round 9 (2018); собрать страновой датасет из World Bank и Transparency International за тот же год.
2. Проверить 10-факторную, 4-факторную плоскую и иерархическую (higher-order) CFA-модели ценностей Шварца на объединённой европейской выборке.
3. Провести тест конфигуральной (configural) и приближённой метрической (metric) инвариантности измерительной модели по 29 странам.
4. Построить серию многоуровневых моделей социального доверия: нулевую, индивидуальный уровень, индивидуальный + страновый уровень. Для финальной спецификации проверить коэффициент инфляции дисперсии (VIF, variance inflation factor) и распределение остатков.
5. Построить серию моделей машинного обучения (линейные модели с регуляризацией, Random Forest - случайный лес, XGBoost - градиентный бустинг) для предсказания индивидуального уровня доверия. Оценить устойчивость рейтинга предикторов через permutation importance (важность, оценённая перестановкой признака) и SHAP-значения (SHapley Additive exPlanations, значения Шепли для вклада признака в предсказание). Провести альтернативную бинарную постановку задачи с анализом калибровки вероятностей и подбора порога классификации.
6. Сопоставить результаты двух подходов (multilevel и ML) через ранговую корреляцию важности признаков и содержательный анализ расхождений.



### 1.5. Научная новизна

Работа одновременно применяет CFA с измерительной инвариантностью, многоуровневую регрессию и ML-модели на одних и тех же данных ESS R9 (2018), формализуя сопоставление двух последних подходов через ранговую корреляцию важности признаков. Большинство прикладных работ по социальному доверию ограничивается одной из этих методологий. Комбинация позволяет использовать факторные оценки (factor scores) из CFA как признаки ML-модели и многоуровневой регрессии одновременно и проверить, насколько итоговый рейтинг предикторов устойчив к выбору подхода.



### 1.6. Практическая значимость

- **Прикладная политика.** Результаты количественно подтверждают, что институциональный фактор (индекс восприятия коррупции, CPI) доминирует над чисто экономическим (валовой внутренний продукт на душу населения, GDP per capita) и распределительным (индекс Джини, Gini) на уровне стран. Для разработчиков политики это аргумент в пользу инвестиций в институциональную среду, а не только в перераспределение доходов.
- **Академическое сообщество.** Кросс-национальный анализ с численным сопоставлением ML и многоуровневой регрессии на одних и тех же признаках даёт материал для последующих работ о том, насколько устойчивы содержательные выводы к выбору методологического подхода.



### 1.7. Исследовательские вопросы

**RQ (research question, исследовательский вопрос) 1.** Воспроизводится ли десятикомпонентная модель базовых ценностей Шварца в объединённой европейской выборке ESS?

**RQ2.** Одинакова ли структура ценностей между странами? Допустимо ли прямое сопоставление ценностных индикаторов (три уровня инвариантности: конфигуральная - configural, метрическая - metric, скалярная - scalar)?

**RQ3.** Какую долю дисперсии социального доверия объясняет страновой уровень (ICC, intraclass correlation coefficient, внутриклассовый коэффициент корреляции)?

**RQ4.** Какие индивидуальные факторы (ценности, медиапотребление, демография) связаны с социальным доверием после контроля на страновые эффекты?

**RQ5.** Как страновые характеристики (GDP, Gini, CPI) связаны с уровнем социального доверия и насколько они улучшают модель сверх индивидуального уровня?



### 1.8. Гипотезы

**H1.** Десятифакторная первичная структура PVQ-21 (Portrait Values Questionnaire, опросник портретов ценностей Шварца, 21 итем) воспроизводится на объединённой выборке ESS9. Критерий соответствия модели: CFI (Comparative Fit Index, сравнительный индекс соответствия) > 0.90 и RMSEA (Root Mean Square Error of Approximation, корень из среднеквадратичной ошибки аппроксимации) < 0.08. 4-факторная модель высших измерений рассматривается как компактная альтернатива.

**H2.** Для 10-факторной модели Шварца достигается конфигуральная и частичная метрическая инвариантность между странами. Скалярная инвариантность формально не тестируется (пакет semopy 2.3.11 не поддерживает constrained multigroup CFA); по литературе (Davidov, Schmidt, Schwartz 2008) в ESS она обычно не достигается.

**H3.** ICC для целевой переменной ppltrst (people trust, доверие к большинству людей; шкала 0-10) находится в диапазоне 0.10-0.20: страновой уровень объясняет значимую, но не доминирующую часть дисперсии индивидуальных оценок доверия.

**H4.** Ценности self-transcendence (самотрансценденция, забота о других) положительно связаны с доверием, conservation (сохранение, ориентация на стабильность) - отрицательно, self-enhancement (самоутверждение, конкуренция за статус) - отрицательно (слабо). Механизм (Uslaner 2002): альтруистические ценности формируют позитивные ожидания в отношении других.

**H5.** Страновые GDP per capita, Gini и CPI связаны с уровнем доверия: выше доход и CPI - выше доверие, выше неравенство - ниже.

## 2. Теоретическая рамка

### 2.1. Базовые ценности Шварца

Теория базовых человеческих ценностей Шварца (Schwartz 1992, 2012) описывает ценности как устойчивые убеждения о желательных целях, выступающие ориентирами в жизни индивида. Десять первичных ценностей:

- **Self-Direction** (самостоятельность) - SD
- **Stimulation** (стимуляция) - ST
- **Hedonism** (гедонизм) - HE
- **Achievement** (достижения) - AC
- **Power** (власть) - PO
- **Security** (безопасность) - SE
- **Conformity** (конформизм) - CO
- **Tradition** (традиция) - TR
- **Benevolence** (благожелательность) - BE
- **Universalism** (универсализм) - UN

Эти ценности образуют циркумплекс: соседние совместимы, противоположные конфликтуют.

Десять ценностей группируются в четыре высших измерения:

- **Openness to Change** (открытость изменениям): Self-Direction + Stimulation + Hedonism. Автономия, новизна, удовольствие.
- **Conservation** (сохранение): Security + Conformity + Tradition. Стабильность, нормы, традиция.
- **Self-Transcendence** (самотрансценденция): Universalism + Benevolence. Забота о других, равенстве, природе.
- **Self-Enhancement** (самоутверждение): Achievement + Power. Личный успех, статус, ресурсы.

В ESS ценности измеряются сокращённой батареей Portrait Values Questionnaire (PVQ-21). Её инвариантность между европейскими странами изучалась в нескольких работах на ESS (Davidov, Schmidt, Schwartz 2008). Типичный результат: конфигуральная инвариантность (общая структура факторов в странах) и частичная метрическая достигаются, скалярная - нет.



### 2.2. Социальное доверие

В литературе разграничивают партикулярное (знакомым), институциональное (государственным институтам) и генерализованное (большинству людей в обществе, Delhey and Newton 2005). В работе под социальным доверием понимается именно генерализованное, измеряемое в ESS шкалой ppltrst от 0 до 10.

Конкурируют две объяснительные схемы. Социально-капитальная (Putnam 2000) связывает доверие с плотностью гражданских ассоциаций и добровольных контактов. Институциональная (Rothstein and Uslaner 2005) видит доверие как функцию качества институтов и экономического неравенства: честные институты и низкое неравенство поднимают доверие, коррумпированные и высокое неравенство - опускают. Кросс-национальные данные в основном поддерживают институциональную перспективу.



### 2.3. Связь ценностей и доверия

Uslaner (2002) описывает генерализованное доверие как моральное ожидание, опирающееся на внутренние ценностные установки. Альтруистические ценности (Self-Transcendence: universalism + benevolence) формируют позитивные ожидания в отношении других и должны коррелировать с доверием положительно. Conservation, напротив, связан с ориентацией на близкий круг и может ограничивать доверие к незнакомцам. Self-Enhancement, связанный с конкурентной мотивацией, теоретически должен давать отрицательный знак. Openness to Change не даёт однозначного направления: любопытство совместимо с доверием, но акцент на личной автономии может ослаблять коллективные моральные нормы.



### 2.4. Многоуровневый подход

Кросс-национальные опросы имеют иерархическую структуру: индивиды вложены в страны. Игнорирование структуры приводит к смешению межстрановых и индивидуальных эффектов и к заниженным стандартным ошибкам из-за кластеризации (Hox 2018).

Решение - многоуровневая (multilevel) регрессия с random intercept (случайный свободный член по стране). Она раскладывает дисперсию зависимой переменной на между-страновую и внутри-страновую, позволяет оценить внутриклассовый коэффициент корреляции (ICC) и корректно включать предикторы обоих уровней. 29 стран - нижняя граница рекомендованного диапазона для частотной оценки параметров на уровне 2, поэтому страновые коэффициенты мы интерпретируем консервативно.



### 2.5. Страновые факторы

На уровне страны мы используем три показателя за 2018 год, покрывающие основные механизмы из литературы о доверии:

- **GDP per capita PPP** (World Bank, `NY.GDP.PCAP.PP.CD`) - валовой внутренний продукт на душу населения по паритету покупательной способности (PPP, purchasing power parity). Характеристика экономического благосостояния.
- **Gini index** (World Bank, `SI.POV.GINI`) - индекс Джини, уровень неравенства доходов.
- **Corruption Perceptions Index** (Transparency International) - индекс восприятия коррупции, экспертная оценка чистоты институтов (0 - высокая коррупция, 100 - низкая).

Этот минимальный набор покрывает экономическую, распределительную и институциональную составляющие. Более широкие наборы (V-Dem, Freedom House, Hofstede) дают дополнительные измерения, но мы их не используем, чтобы удержать объём работы в рамках магистерской.



### 2.6. ML как альтернативный методологический подход

Помимо многоуровневой регрессии мы применяем модели машинного обучения (Random Forest, XGBoost) к той же задаче: предсказание индивидуального уровня ppltrst по тому же набору из 15 признаков. Для интерпретации используются permutation importance (оценка важности признака по падению качества модели R² при случайной перестановке его значений) и SHAP-значения (Shapley values, вклад каждого признака в конкретное предсказание; Lundberg and Lee 2017).

Многоуровневая регрессия и ML отвечают на разные вопросы. Многоуровневая модель даёт параметрическую оценку с понятной интерпретацией коэффициентов и учётом иерархической структуры. ML-модели не делают допущений о линейной форме связи и могут учитывать нелинейные зависимости и взаимодействия между признаками. Если оба подхода независимо выделяют одни и те же ведущие предикторы, это говорит о том, что содержательные выводы устойчивы к выбору методологии.



## 3. Данные

Используются два источника: индивидуальные данные European Social Survey Round 9 (2018) и страновые индикаторы за 2018 год из World Bank и Transparency International.



### 3.1. ESS Round 9 (2018)

ESS - кросс-национальное исследование, проводимое раз в два года по единой методологии. Данные собираются личным интервью по стратифицированной случайной выборке в каждой стране. Round 9 выбран по нескольким причинам:

- Данные 2018 года достаточно свежие для ВКР 2026 года.
- Round 10 (2020-2022) затронут ковидной модификацией методологии: в ряде стран PVQ был перенесён в self-completion часть, а интегрированный файл полный набор PVQ-итемов не содержит.
- Round 9 покрывает 29 европейских стран и содержит полную PVQ-21 для измерения ценностей по Шварцу.

Исходный файл ESS9 integrated file edition 3.3 (`ESS9e03_3.csv`) содержит 49519 респондентов и 575 переменных.



### 3.2. Отбор переменных

Из 575 переменных ESS отобрано 37 релевантных:

**Целевая переменная:** `ppltrst` - генерализованное социальное доверие (0-10).

**Ценности (PVQ-21, 21 итем):** ipcrtiv, imprich, ipeqopt, ipshabt, impsafe, impdiff, ipfrule, ipudrst, ipmodst, ipgdtim, impfree, iphlppl, ipsuces, ipstrgv, ipadvnt, ipbhprp, iprspot, iplylfr, impenv, imptrad, impfun. Каждый итем - описание гипотетического человека, оцениваемое по шкале от 1 (очень похож на меня) до 6 (совсем не похож).

**Медиапотребление (3 переменные):** `netusoft` (частота использования интернета, 1-5), `netustm` (минуты в интернете в день), `nwspol` (минуты политических новостей в день). В ESS R9 блок вопросов о ТВ и радио сокращён по сравнению с ранними раундами, это фиксируем как ограничение.

**Контроли (демография):** `agea` (возраст), `gndr` (пол), `eduyrs` (годы образования), `eisced` (уровень образования ISCED), `hinctnta` (дециль дохода домохозяйства), `domicil` (тип населённого пункта).

### 3.4. Обработка пропусков и трансформации

В ESS специальные коды для пропусков лежат за пределами валидных значений, но близко к ним по разряду. Для шкалы 1-6 это 7 (Refusal), 8 (Don't know), 9 (No answer); для шкалы 0-10 - 77/88/99; для переменных в минутах - 6666/7777/8888/9999. Замена проводится явно по группам переменных: для каждой группы прописан свой набор кодов. Такой подход точнее, чем общий каскад по числу разрядов, поскольку не рискует затронуть валидные крайние значения.

Верхняя обрезка экстремальных значений (top-coding):

- `eduyrs` > 25 лет - значение обрезается до 25 (0.45% наблюдений с явными ошибками ввода).
- `netustm`, `nwspol` > 1440 минут (больше суток) - заменяется на NaN.

Для правосторонне асимметричных переменных (`netustm`, `nwspol`) применена логарифмическая трансформация `log(1 + x)`. Для переменных с существенной долей пропусков, критически важных для модели (`netustm` 30.4%, `hinctnta` 19.5%), применена медианная замена пропусков (median imputation) внутри страны.

### 4.4. Предварительные связи ценностей с доверием

Строим простые mean-индексы четырёх высших измерений Шварца. Шкала PVQ инвертирована: 1 = "очень похож на меня", 6 = "совсем не похож". Чтобы высокий индекс означал большую важность ценности, берём среднее по итемам и инвертируем знак.


```python
# Простые mean-индексы четырёх высших измерений (для предварительного EDA)
ES_OPENNESS = ['ipcrtiv', 'impfree', 'impdiff', 'ipadvnt', 'impfun', 'ipgdtim']
ES_CONSERVATION = ['impsafe', 'ipstrgv', 'ipfrule', 'ipbhprp', 'ipmodst', 'imptrad']
ES_SELF_TRANS = ['iplylfr', 'iphlppl', 'ipeqopt', 'ipudrst', 'impenv']
ES_SELF_ENH = ['ipsuces', 'ipshabt', 'imprich', 'iprspot']

tmp = df.copy()
tmp['openness_m'] = -tmp[ES_OPENNESS].mean(axis=1)
tmp['conservation_m'] = -tmp[ES_CONSERVATION].mean(axis=1)
tmp['self_trans_m'] = -tmp[ES_SELF_TRANS].mean(axis=1)
tmp['self_enh_m'] = -tmp[ES_SELF_ENH].mean(axis=1)

print('Spearman-корреляция простых mean-индексов ценностей и ppltrst:')
val_cols = ['openness_m', 'conservation_m', 'self_trans_m', 'self_enh_m']
corr_val = tmp[['ppltrst'] + val_cols].corr(method='spearman')['ppltrst'].drop('ppltrst').round(3)
display(corr_val.to_frame('corr_with_ppltrst'))
```


Self-transcendence даёт положительную связь с доверием, conservation и self-enhancement - отрицательную. Направления соответствуют теории Uslaner (2002) и гипотезе H4. Формальная проверка с контролем на остальные переменные - в разделе 6. Абсолютные значения Spearman ρ здесь небольшие (0.05-0.11), но это ожидаемо для одиночных корреляций на большой выборке; в мультивариантных моделях раздела 6 эффекты становятся сильнее после учёта демографических контролей.



## 5. Часть 1. Структура базовых ценностей

Часть посвящена проверке измерительной модели Шварца на данных ESS R9. Оцениваем три конкурирующих CFA-спецификации на объединённой выборке, затем тестируем инвариантность структуры между странами. Результаты определяют, как ценности входят в модели Части 2.


### 5.1. Выборка для CFA


```python
# Подвыборка для CFA: respondents без пропусков ни по одному из 21 PVQ-итема (listwise)
df_cfa = df.dropna(subset=PVQ_ITEMS).copy()

print(f'Полная выборка ESS R9: {len(df):,}')
print(f'Listwise по 21 PVQ items: {len(df_cfa):,} ({len(df_cfa) / len(df) * 100:.1f}% от исходной)')
print(f'Потеря: {len(df) - len(df_cfa):,} респондентов с хотя бы одним пропуском PVQ')
```


Построчное удаление наблюдений с пропусками (listwise deletion) снижает выборку примерно на 8%. Пропуски PVQ сосредоточены у малого числа респондентов (1-2 пропущенных итема на человека), поэтому альтернативный метод FIML (full information maximum likelihood, полноинформационный метод максимального правдоподобия, использующий все доступные значения) дал бы небольшой прирост выборки при существенно большей сложности реализации. Мы используем listwise.



### 5.2. Спецификация моделей

PVQ-21 - 6-балльная шкала ("совсем не похож" ... "очень похож"). Мы трактуем её как континуальную и оцениваем CFA методом максимального правдоподобия на корреляционной матрице Пирсона. Такое приближение допустимо при 5 и более категориях и приблизительно симметричных распределениях, что подтверждается описательной статистикой PVQ-итемов из раздела 4.

Мы сравниваем три модели. Синтаксис semopy: `Фактор =~ итем1 + итем2` означает, что факторная переменная формируется из перечисленных итемов.

- **10-factor Schwartz** (десятифакторная модель Шварца) - десять первичных ценностей, каждая с 2-3 итемами.
- **Higher-order** (иерархическая модель) - те же 10 ценностей, сгруппированные в 4 высших измерения (Openness, Conservation, Self-Transcendence, Self-Enhancement).
- **4-factor flat** (плоская четырёхфакторная модель) - 21 итем напрямую грузится на 4 высших измерения, минуя уровень первичных ценностей.


```python
from semopy import Model, calc_stats

MODEL_10 = '''
SD =~ ipcrtiv + impfree
ST =~ impdiff + ipadvnt
HE =~ impfun + ipgdtim
AC =~ ipsuces + ipshabt
PO =~ imprich + iprspot
SE =~ impsafe + ipstrgv
CO =~ ipfrule + ipbhprp
TR =~ ipmodst + imptrad
BE =~ iplylfr + iphlppl
UN =~ ipeqopt + ipudrst + impenv
'''

MODEL_HO = '''
SD =~ ipcrtiv + impfree
ST =~ impdiff + ipadvnt
HE =~ impfun + ipgdtim
AC =~ ipsuces + ipshabt
PO =~ imprich + iprspot
SE =~ impsafe + ipstrgv
CO =~ ipfrule + ipbhprp
TR =~ ipmodst + imptrad
BE =~ iplylfr + iphlppl
UN =~ ipeqopt + ipudrst + impenv

Openness =~ SD + ST + HE
Conservation =~ SE + CO + TR
SelfTrans =~ UN + BE
SelfEnh =~ AC + PO
'''

MODEL_4 = '''
Openness =~ ipcrtiv + impfree + impdiff + ipadvnt + impfun + ipgdtim
Conservation =~ impsafe + ipstrgv + ipfrule + ipbhprp + ipmodst + imptrad
SelfTrans =~ iplylfr + iphlppl + ipeqopt + ipudrst + impenv
SelfEnh =~ ipsuces + ipshabt + imprich + iprspot
'''
```


Считаем SRMR (standardized root mean square residual) вручную: semopy в версии 2.3.11 этот индекс не возвращает, поэтому берём observed-ковариацию из данных, model-implied ковариацию из обученной модели, стандартизуем разность и усредняем квадраты по нижнему треугольнику. Формула из документации lavaan.


```python
def compute_srmr(m, data):
    obs_vars = m.vars['observed']
    S = np.cov(data[obs_vars].to_numpy(), rowvar=False, ddof=0)
    Sigma = m.calc_sigma()[0]
    sd_obs = np.sqrt(np.diag(S))
    resid = (S - Sigma) / np.outer(sd_obs, sd_obs)
    mask = np.tril(np.ones_like(resid, dtype=bool))
    vals = resid[mask]
    return float(np.sqrt((vals ** 2).sum() / len(vals)))

def fit_and_summarise(desc, data):
    m = Model(desc)
    m.fit(data, obj='MLW')
    stats = calc_stats(m)
    stats['SRMR'] = compute_srmr(m, data)
    return m, stats

# Оцениваем три модели
m10, stats10 = fit_and_summarise(MODEL_10, df_cfa[PVQ_ITEMS])
mho, statsho = fit_and_summarise(MODEL_HO, df_cfa[PVQ_ITEMS])
m4, stats4 = fit_and_summarise(MODEL_4, df_cfa[PVQ_ITEMS])

# Сводная таблица
stats10.index = ['10-factor Schwartz']
statsho.index = ['higher-order 10->4']
stats4.index = ['4-factor flat']
fit = pd.concat([stats10, statsho, stats4])
fit['chi2_over_df'] = fit['chi2'] / fit['DoF']
display(fit[['chi2', 'DoF', 'chi2_over_df', 'CFI', 'TLI', 'RMSEA', 'SRMR', 'AIC', 'BIC']].round(3))

print()
print('Стандартные критерии приемлемого соответствия (Hu & Bentler 1999, Chen 2007):')
print('  CFI >= 0.90 приемлемо, >= 0.95 хорошо')
print('  RMSEA <= 0.08 приемлемо, <= 0.06 хорошо')
print('  SRMR <= 0.08 приемлемо')
```


10-факторная модель даёт приемлемый fit по всем трём основным индексам (CFI = 0.906, RMSEA = 0.058, SRMR = 0.050) и проходит стандартные пороги. Higher-order 10→4 и 4-factor flat показывают более слабый fit (CFI 0.85-0.87): жёсткое объединение итемов в четыре измерения теряет точность, характерную для первичной структуры.

Для интерпретации: десять первичных ценностей - более точное представление данных, чем четыре высших измерения. При этом 10 сильно коррелирующих предикторов неэффективно использовать в многоуровневой модели, поэтому в Части 2 мы агрегируем их до 4 композитных индексов (усреднение факторных оценок соответствующих первичных ценностей).

**Итог по H1.** Первичная 10-факторная структура Шварца воспроизводится (CFI > 0.90, RMSEA < 0.06, SRMR = 0.05). H1 подтверждается.

### 5.4. Configural invariance: структура по странам

Configural invariance (конфигуральная инвариантность) означает, что одна и та же структура факторов (какие итемы какому фактору принадлежат) воспроизводится в каждой стране. Для проверки оцениваем 10-факторную CFA независимо в каждой из 29 стран через `semopy.multigroup.multigroup`.


```python
from semopy.multigroup import multigroup

# Маппинг ISO-кодов ESS в русские названия стран (используется для таблицы)
COUNTRY_NAMES = {
    'AT': 'Австрия', 'BE': 'Бельгия', 'BG': 'Болгария', 'CH': 'Швейцария',
    'CY': 'Кипр', 'CZ': 'Чехия', 'DE': 'Германия', 'DK': 'Дания',
    'EE': 'Эстония', 'ES': 'Испания', 'FI': 'Финляндия', 'FR': 'Франция',
    'GB': 'Великобритания', 'HR': 'Хорватия', 'HU': 'Венгрия', 'IE': 'Ирландия',
    'IS': 'Исландия', 'IT': 'Италия', 'LT': 'Литва', 'LV': 'Латвия',
    'ME': 'Черногория', 'NL': 'Нидерланды', 'NO': 'Норвегия', 'PL': 'Польша',
    'PT': 'Португалия', 'RS': 'Сербия', 'SE': 'Швеция', 'SI': 'Словения',
    'SK': 'Словакия',
}

# Независимая оценка CFA в каждой стране (configural)
mg = multigroup(MODEL_10, df_cfa[['cntry'] + PVQ_ITEMS], group='cntry')

rows = []
for c in mg.groups:
    s = mg.stats[c].iloc[0]
    rows.append({
        'country_code': c,
        'country': COUNTRY_NAMES.get(c, c),
        'n': mg.n_obs[c],
        'CFI': float(s.get('CFI', np.nan)),
        'TLI': float(s.get('TLI', np.nan)),
        'RMSEA': float(s.get('RMSEA', np.nan)),
    })
fit_country = pd.DataFrame(rows).sort_values('CFI', ascending=False).reset_index(drop=True)
```

```python
# График CFI по странам с референсными порогами
fig, ax = plt.subplots(figsize=(12, 4))
ax.bar(fit_country['country_code'], fit_country['CFI'], color='steelblue')
ax.axhline(0.90, color='green', linestyle='--', label='CFI = 0.90 (приемлемо)')
ax.axhline(0.95, color='darkgreen', linestyle=':', label='CFI = 0.95 (хорошо)')
ax.set_ylabel('CFI'); ax.set_xlabel('Страна (ISO-код)')
ax.set_ylim(0.80, 1.0)
ax.set_title('Configural invariance: CFI 10-факторной CFA по странам')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout(); plt.show()

n_total = len(fit_country)
n_90 = (fit_country['CFI'] >= 0.90).sum()
n_85 = (fit_country['CFI'] >= 0.85).sum()
n_rmsea_ok = (fit_country['RMSEA'] <= 0.08).sum()
print(f'Стран с CFI >= 0.90: {n_90}/{n_total}')
print(f'Стран с CFI >= 0.85: {n_85}/{n_total}')
print(f'Стран с RMSEA <= 0.08: {n_rmsea_ok}/{n_total}')
print()
print('5 стран с худшим CFI:')
display(fit_country.tail(5)[['country_code', 'country', 'n', 'CFI', 'TLI', 'RMSEA']].round(3))
```


Структура воспроизводится в 28 странах из 29 с CFI >= 0.85. Строгий порог CFI = 0.90 достигается в 7 странах, остальные лежат в зоне 0.85-0.90. Единственная страна с маргинальным результатом - Словения (CFI = 0.840). RMSEA не превышает 0.08 в 25 странах. Такой уровень согласия считается приемлемым для прикладного кросс-национального анализа (Davidov et al. 2008). **Конфигуральная инвариантность подтверждается.**



### 5.5. Приближённая метрическая инвариантность

Метрическая инвариантность (metric invariance) означает, что факторные нагрузки (loadings - коэффициенты, с которыми каждый итем связан со скрытой ценностью) одинаковы между странами. При её выполнении коэффициенты регрессии ценностей на доверие между странами сопоставимы.

Полный тест metric invariance в multigroup framework требует оценки моделей с ограничениями на равенство нагрузок; пакет semopy 2.3.11 это не поддерживает. Мы используем упрощённую версию: собираем нагрузки из configural-оценки (раздел 5.4) и смотрим, насколько они варьируют между странами. Если разброс мал, metric invariance выполняется приближённо.


```python
# Собираем все measurement loadings из configural-оценки в одну таблицу
all_params_rows = []
for c in mg.groups:
    est = mg.estimates[c].copy()
    est['country'] = c
    all_params_rows.append(est)
all_params = pd.concat(all_params_rows, ignore_index=True)

load = all_params[all_params['op'] == '~'].copy()
load['Estimate'] = pd.to_numeric(load['Estimate'], errors='coerce')
load['key'] = load['rval'] + ' -> ' + load['lval']

# Статистика вариабельности по каждому пути фактор -> итем
lv = (load.groupby(['rval', 'lval'])['Estimate']
      .agg(['mean', 'std', 'min', 'max', 'count']).reset_index()
      .rename(columns={'rval': 'factor', 'lval': 'item'}))
lv['range'] = lv['max'] - lv['min']
# Исключаем marker-itемы (std ≈ 0, так как они фиксированы на 1.0)
lv = lv[lv['std'] > 1e-6].sort_values('std').reset_index(drop=True)
lv['key'] = lv['factor'] + ' -> ' + lv['item']
display(lv[['factor', 'item', 'mean', 'std', 'min', 'max', 'range']].round(3))
```

```python
# Boxplot свободно оцениваемых нагрузок между 29 странами
free = load[load.groupby('key')['Estimate'].transform('std') > 1e-6].copy()
order = lv['key'].tolist()
data_box = [free.loc[free['key'] == k, 'Estimate'].dropna().values for k in order]

fig, ax = plt.subplots(figsize=(11, 5))
ax.boxplot(data_box, tick_labels=order, vert=True)
ax.axhline(1.0, color='red', linestyle='--', alpha=0.5, label='Marker-итемы (фиксированы на 1.0)')
ax.set_ylabel('Factor loading')
ax.set_title('Вариабельность факторных нагрузок между 29 странами')
plt.xticks(rotation=45, ha='right')
ax.legend()
plt.tight_layout(); plt.show()

n_stable = (lv['std'] < 0.15).sum()
n_problem = (lv['std'] >= 0.25).sum()
print(f'Всего свободно оцениваемых loadings: {len(lv)}')
print(f'Стабильные (SD < 0.15): {n_stable}')
print(f'Проблемные (SD >= 0.25): {n_problem}')
if n_problem > 0:
    print()
    print('Проблемные items:')
    display(lv[lv['std'] >= 0.25][['factor', 'item', 'std']].round(3))
```


Пять из одиннадцати свободно оцениваемых нагрузок устойчивы между странами (SD < 0.15). Два итема показывают большой разброс нагрузок между странами: CO → ipbhprp (SD = 0.37) и PO → iprspot (SD = 0.64 с диапазоном 0.22-3.22). Это означает, что содержательный смысл этих итемов в разных культурах различается, и прямое сравнение средних значений Conformity и Power между странами было бы некорректным. Для регрессионных коэффициентов Части 2 достигнутой частичной метрической инвариантности достаточно.

Формальный тест скалярной инвариантности в multigroup framework пакет semopy 2.3.11 не реализует, поэтому мы ограничиваемся оценкой вариабельности нагрузок - приближённой метрической инвариантностью. **Частичная метрическая инвариантность подтверждается.** Скалярная инвариантность (равенство интерсептов между странами) формально не тестировалась; по литературе (Davidov et al. 2008) она в ESS обычно не достигается. Для наших задач это не критично: нас интересуют коэффициенты регрессии, а не средние значения ценностей по странам.



### 5.6. Итог Части 1

- **H1 подтверждается.** 10-факторная первичная модель Шварца воспроизводится на данных ESS9 (CFI = 0.906, RMSEA = 0.058, SRMR = 0.050).
- **H2 подтверждается частично.** Конфигуральная инвариантность: 28 из 29 стран с CFI >= 0.85. Частичная метрическая инвариантность: 9 из 11 нагрузок устойчивы, 2 итема (`ipbhprp`, `iprspot`) показывают культурную специфику. Скалярная инвариантность формально не проверена.
- **Содержательное следствие.** Коэффициенты регрессии ценностей на доверие между странами корректно сопоставимы. Прямое сравнение средних значений ценностей между странами не используется в Части 2.
- **Методологическое следствие.** В Часть 2 ценности входят как 4 композитных индекса, полученных усреднением factor scores из 10-факторной CFA (Openness = mean(SD, ST, HE), Conservation = mean(SE, CO, TR), Self-Transcendence = mean(UN, BE), Self-Enhancement = mean(AC, PO)).



## 6. Часть 2. Моделирование социального доверия

### 6.0. Стратегия двух подходов

К одной и той же задаче (предсказание и объяснение индивидуального уровня ppltrst по 15 признакам: 4 композитных ценности + 3 медиапеременные + 5 демографических + 3 страновых; gndr закодирован как `female` = 1/0) мы применяем два независимых методологических подхода.

**ML-подход (6.1)** отвечает на вопрос "какие признаки полезно знать для предсказания уровня доверия у нового респондента". Используются линейная регрессия (OLS, ordinary least squares, метод наименьших квадратов), Ridge (гребневая регрессия с L2-регуляризацией), Lasso (регрессия с L1-регуляризацией и отбором признаков), Random Forest и XGBoost. Гиперпараметры подбираются через GridSearchCV (grid search с кросс-валидацией, поиск оптимальной комбинации по сетке значений). Выборка делится 80/20 на train (обучающую) и hold-out test (отложенную тестовую, на которой модель не обучается). Для интерпретации используются permutation importance и SHAP summary plot (сводная диаграмма вкладов признаков). Бинарная постановка (ppltrst >= 5) рассматривается отдельно: график калибровки, Brier score (мера точности вероятностных предсказаний, чем ниже тем лучше) и подбор порога классификации.

**Многоуровневый подход (6.2)** отвечает на вопрос "какова структура эффектов индивидуальных и страновых предикторов с учётом вложенности индивидов в страны". Оцениваются три вложенные спецификации (нулевая, индивидуальные, индивидуальные + страновые), диагностика VIF и QQ-plot остатков (quantile-quantile plot - график соответствия квантилей остатков нормальному распределению).

**Сопоставление (6.3)** формализовано через Spearman ρ (коэффициент ранговой корреляции Спирмена) между рангами важности признаков в ML и многоуровневой регрессии. Это даёт численную оценку того, насколько выводы устойчивы к выбору подхода.



### 6.1. ML-подход

#### 6.1.1. Выбор метрик

Для регрессии мы выбираем R² (коэффициент детерминации, доля объяснённой дисперсии), MAE (mean absolute error, средняя абсолютная ошибка - устойчива к выбросам на шкале 0-10) и RMSE (root mean square error, корень из среднеквадратичной ошибки - штрафует большие ошибки сильнее).

Для классификации с дисбалансом 65/35 F1 мажоритарного класса даёт завышенную картину даже у DummyClassifier (тривиальная модель-заглушка, всегда предсказывающая один и тот же класс). Честные метрики - F1_macro (невзвешенное среднее F1 по обоим классам), ROC-AUC (площадь под ROC-кривой, area under the receiver operating characteristic curve - не зависит от выбранного порога классификации) и Brier score (средняя квадратичная ошибка предсказанной вероятности, мера калибровки). На эти три метрики и ориентируемся.



#### Подготовка признаков для Части 2

Четыре композитных индекса ценностей строим из факторных оценок 10-факторной CFA (раздел 5). Для каждого высшего измерения берём среднее факторных оценок соответствующих первичных ценностей и инвертируем знак (PVQ-шкала "1 = очень похож" требует инверсии, чтобы высокий индекс означал высокую важность ценности). Для netustm и hinctnta - медианная замена пропусков внутри страны. Всё подготавливается один раз и используется для ML и multilevel.


```python
# 1. Извлекаем факторные оценки из m10 (10-factor CFA)
fs10 = m10.predict_factors(df_cfa[PVQ_ITEMS])
fs10.index = df_cfa.index
fs10.columns = [f'fs_{c}' for c in fs10.columns]

# 2. Собираем композитные индексы ценностей (с инверсией знака)
fs10['openness'] = -fs10[['fs_SD', 'fs_ST', 'fs_HE']].mean(axis=1)
fs10['conservation'] = -fs10[['fs_SE', 'fs_CO', 'fs_TR']].mean(axis=1)
fs10['self_trans'] = -fs10[['fs_UN', 'fs_BE']].mean(axis=1)
fs10['self_enh'] = -fs10[['fs_AC', 'fs_PO']].mean(axis=1)
values_cols = ['openness', 'conservation', 'self_trans', 'self_enh']

# 3. Merge факторных оценок в основной датасет (по индексу) + country-level
analytic = df.merge(fs10[values_cols], left_index=True, right_index=True, how='left')
analytic = analytic.merge(country_level[['cntry', 'gdp_pcap_ppp', 'gini', 'cpi']],
                          on='cntry', how='left')

# 4. Медианная замена пропусков внутри страны для hinctnta и netustm
analytic['hinctnta'] = analytic.groupby('cntry')['hinctnta'].transform(lambda s: s.fillna(s.median()))
analytic['netustm'] = analytic.groupby('cntry')['netustm'].transform(lambda s: s.fillna(s.median()))

# 5. Дополнительные преобразования
analytic['female'] = (analytic['gndr'] == 2).astype(int)
analytic['log_netustm'] = np.log1p(analytic['netustm'])
analytic['log_nwspol'] = np.log1p(analytic['nwspol'])

# 6. Аналитическая выборка: listwise по всем 15 признакам + ppltrst
core_vars = (['ppltrst'] + values_cols
             + ['netusoft', 'log_netustm', 'log_nwspol',
                'agea', 'female', 'eduyrs', 'hinctnta', 'domicil',
                'gdp_pcap_ppp', 'gini', 'cpi'])
sample = analytic.dropna(subset=core_vars).copy()

# 7. Z-стандартизация непрерывных предикторов (для сопоставимости коэффициентов)
for v in (values_cols + ['netusoft', 'log_netustm', 'log_nwspol',
                         'agea', 'eduyrs', 'hinctnta', 'domicil',
                         'gdp_pcap_ppp', 'gini', 'cpi']):
    sample[v + '_z'] = (sample[v] - sample[v].mean()) / sample[v].std()

print(f'Аналитическая выборка: {len(sample):,} респондентов из {len(df):,} ({len(sample)/len(df)*100:.1f}%)')
print(f'Стран: {sample["cntry"].nunique()}, композитных ценностей: {len(values_cols)}')
```


#### 6.1.2. Базовые линейные модели и нелинейные с подбором гиперпараметров

Сначала split train/test 80/20 со стратификацией по стране (каждая страна представлена в обеих выборках пропорционально своему размеру). Затем оцениваем 6 регрессионных моделей: Dummy (предсказывает среднее), OLS, Ridge с подбором alpha, Lasso с подбором alpha, Random Forest и XGBoost с полным GridSearchCV.


```python
import contextlib, io, sys, os

# Контекстный менеджер, перехватывающий stderr из worker-процессов joblib.
# sklearn 1.8 генерирует UserWarning из каждого joblib-worker при n_jobs=-1.
# Родительские warnings.filterwarnings в workers не наследуются, поэтому глушим весь stderr
# только вокруг .fit()-вызовов GridSearchCV.
@contextlib.contextmanager
def suppress_stderr():
    save = sys.stderr
    sys.stderr = io.StringIO()
    # дублируем через файловый дескриптор (joblib пишет напрямую в fd=2)
    saved_fd = os.dup(2)
    devnull = os.open(os.devnull, os.O_WRONLY)
    os.dup2(devnull, 2)
    try:
        yield
    finally:
        os.dup2(saved_fd, 2)
        os.close(devnull)
        os.close(saved_fd)
        sys.stderr = save

from sklearn.model_selection import train_test_split, KFold, cross_validate, StratifiedKFold, GridSearchCV
from sklearn.linear_model import LinearRegression, LogisticRegression, RidgeCV, LassoCV
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.dummy import DummyRegressor, DummyClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error, r2_score,
    f1_score, accuracy_score, roc_auc_score, confusion_matrix,
    brier_score_loss, roc_curve,
)
from sklearn.inspection import permutation_importance
from sklearn.calibration import calibration_curve
from xgboost import XGBRegressor, XGBClassifier

RANDOM_STATE = 42

FEATURES = [
    'openness', 'conservation', 'self_trans', 'self_enh',
    'netusoft', 'log_netustm', 'log_nwspol',
    'agea', 'female', 'eduyrs', 'hinctnta', 'domicil',
    'gdp_pcap_ppp', 'gini', 'cpi',
]

X = sample[FEATURES]
y_cont = sample['ppltrst'].astype(float)
y_bin = (sample['ppltrst'] >= 5).astype(int)
groups = sample['cntry']

X_tr, X_te, y_tr, y_te = train_test_split(
    X, y_cont, test_size=0.2, stratify=groups, random_state=RANDOM_STATE
)
print(f'Train: {len(X_tr):,}, Test: {len(X_te):,}')

cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
```

```python
# GridSearchCV: Random Forest (12 комбинаций x 5 fold = 60 fit-ов)
rf_grid = {
    'n_estimators': [200, 400],
    'max_depth': [8, 12, None],
    'min_samples_leaf': [1, 5],
}
rf_gs = GridSearchCV(
    RandomForestRegressor(n_jobs=-1, random_state=RANDOM_STATE),
    rf_grid, scoring='r2', cv=cv, n_jobs=-1, refit=True,
)
with suppress_stderr():
    rf_gs.fit(X_tr, y_tr)
print(f'Random Forest best params: {rf_gs.best_params_}')
print(f'Random Forest best CV R²:  {rf_gs.best_score_:.4f}')
```

```python
# GridSearchCV: XGBoost (16 комбинаций x 5 fold = 80 fit-ов, tree_method='hist' для скорости)
xgb_grid = {
    'n_estimators': [200, 400],
    'max_depth': [4, 6],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0],
}
xgb_gs = GridSearchCV(
    XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1, verbosity=0, tree_method='hist'),
    xgb_grid, scoring='r2', cv=cv, n_jobs=-1, refit=True,
)
with suppress_stderr():
    xgb_gs.fit(X_tr, y_tr)
print(f'XGBoost best params: {xgb_gs.best_params_}')
print(f'XGBoost best CV R²:  {xgb_gs.best_score_:.4f}')
```

```python
# Сравнение всех 6 регрессионных моделей: train R², test R², gap + CV-статистика
ridge = Pipeline([('sc', StandardScaler()), ('est', RidgeCV(alphas=np.logspace(-2, 2, 9)))])
lasso = Pipeline([('sc', StandardScaler()),
                  ('est', LassoCV(alphas=np.logspace(-3, 1, 9), cv=cv,
                                  random_state=RANDOM_STATE, max_iter=10000))])

models_reg = {
    'DummyRegressor (mean)': DummyRegressor(strategy='mean'),
    'Linear Regression (OLS)': LinearRegression(),
    'Ridge (CV alpha)': ridge,
    'Lasso (CV alpha)': lasso,
    'Random Forest (tuned)': rf_gs.best_estimator_,
    'XGBoost (tuned)': xgb_gs.best_estimator_,
}

scoring = {'R2': 'r2', 'MAE': 'neg_mean_absolute_error', 'RMSE': 'neg_root_mean_squared_error'}

rows = []
fitted_reg = {}
for name, mdl in models_reg.items():
    with suppress_stderr():
        cv_res = cross_validate(mdl, X_tr, y_tr, cv=cv, scoring=scoring, n_jobs=-1, return_train_score=True)
        mdl.fit(X_tr, y_tr)
    y_pred_tr = mdl.predict(X_tr)
    y_pred_te = mdl.predict(X_te)
    row = {
        'model': name,
        'train_R2': r2_score(y_tr, y_pred_tr),
        'test_R2': r2_score(y_te, y_pred_te),
        'cv_R2_mean': cv_res['test_R2'].mean(),
        'cv_R2_std': cv_res['test_R2'].std(),
        'test_MAE': mean_absolute_error(y_te, y_pred_te),
        'test_RMSE': float(np.sqrt(mean_squared_error(y_te, y_pred_te))),
    }
    row['gap_R2'] = row['train_R2'] - row['test_R2']
    rows.append(row)
    fitted_reg[name] = mdl

reg_results = pd.DataFrame(rows).round(4)
display(reg_results[['model', 'train_R2', 'test_R2', 'gap_R2',
                     'cv_R2_mean', 'cv_R2_std', 'test_MAE', 'test_RMSE']])
```


Результаты:

- **DummyRegressor** (тривиальный базовый регрессор, предсказывающий среднее) и **OLS**: gap (разрыв между train R² и test R²) около 0.006 - переобучения (overfit) нет, но и сигнал ограничен (R² = 0.14).
- **Ridge** и **Lasso** дают почти то же, что OLS. Регуляризация не меняет картину: мультиколлинеарность признаков в нашем наборе не критична (VIF на индивидуальном уровне 1-5, см. раздел 6.2).
- **Random Forest** после подбора гиперпараметров: train R² = 0.39, test R² = 0.19, gap = 0.21. Значительный разрыв - признак того, что модель подгоняется под тонкие нелинейности обучающей выборки, которые не переносятся на тестовую.
- **XGBoost** после подбора гиперпараметров: train R² = 0.32, test R² = 0.19, gap = 0.14. Умеренное переобучение, лучше Random Forest за счёт L2-регуляризации и ограниченной скорости обучения (learning rate). Лучший test R² среди всех моделей.

Общий вывод: нелинейные модели дают прирост test R² на 0.04-0.05 над OLS (0.19 против 0.14). Ценой - более существенное переобучение и потеря интерпретируемости. В абсолютных числах уровень предсказания умеренный: доверие - многокомпонентная переменная, и 15 доступных признаков объясняют около 19% её дисперсии.


#### 6.1.3. Permutation importance


```python
# Permutation importance (n_repeats=10): измеряем, насколько падает R² на test-выборке
# при случайной перестановке значений каждого признака. Высокое падение = признак важный.
def perm_importance_table(model, X_te, y_te):
    with suppress_stderr():
        perm = permutation_importance(model, X_te, y_te,
                                       n_repeats=10, n_jobs=-1, random_state=RANDOM_STATE)
    return pd.DataFrame({
        'feature': FEATURES,
        'importance_mean': perm.importances_mean,
        'importance_std': perm.importances_std,
    }).sort_values('importance_mean', ascending=False).round(5)

rf_imp = perm_importance_table(fitted_reg['Random Forest (tuned)'], X_te, y_te)
xgb_imp = perm_importance_table(fitted_reg['XGBoost (tuned)'], X_te, y_te)

print('Random Forest permutation importance:')
display(rf_imp)
print()
print('XGBoost permutation importance:')
display(xgb_imp)
```

```python
# График permutation importance для Random Forest (mean ± std)
ord_df = rf_imp.sort_values('importance_mean')
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(ord_df['feature'], ord_df['importance_mean'],
        xerr=ord_df['importance_std'].values,
        color='steelblue', error_kw=dict(ecolor='black', lw=0.8, capsize=2))
ax.set_xlabel('Снижение R² при перемешивании признака (mean ± std, n_repeats=10)')
ax.set_title('Permutation importance, Random Forest Regressor (test set)')
plt.tight_layout(); plt.show()
```


В обеих моделях CPI занимает первое место с большим отрывом: перестановка значений CPI снижает test R² на 13-16 процентных пунктов. Следом идут GDP per capita, eduyrs, gini, hinctnta, agea. У XGBoost ценностные признаки (self_trans, conservation, self_enh) попадают в середину списка; у RF они ниже.

Рядом с каждым столбиком показано стандартное отклонение по 10 повторам перемешивания - индикатор устойчивости оценки. У ведущих признаков оно мало по сравнению со средним, у нижней трети рейтинга - сопоставимо со средним, то есть порядок признаков в хвосте менее надёжен.



#### 6.1.4. SHAP summary plot

SHAP summary plot (сводная диаграмма распределения вкладов признаков в предсказания) использует формат beeswarm (пчелиный рой): каждая точка - один респондент, горизонтальная позиция - вклад признака в его предсказание.


```python
import shap

# TreeExplainer строит Shapley-значения на основе структуры деревьев XGBoost (быстро).
# Берём подвыборку 2000 объектов из test для скорости отрисовки.
rng = np.random.default_rng(RANDOM_STATE)
idx = rng.choice(len(X_te), size=min(2000, len(X_te)), replace=False)
X_shap = X_te.iloc[idx]

explainer = shap.TreeExplainer(fitted_reg['XGBoost (tuned)'])
sv = explainer.shap_values(X_shap)

# Summary plot (beeswarm)
plt.figure(figsize=(10, 8))
shap.summary_plot(sv, X_shap, feature_names=FEATURES, show=False)
plt.tight_layout(); plt.show()

# Таблица mean |SHAP|
shap_df = pd.DataFrame({
    'feature': FEATURES,
    'mean_abs_shap': np.abs(sv).mean(axis=0),
}).sort_values('mean_abs_shap', ascending=False).round(5)
print('Mean |SHAP| (среднее абсолютное влияние на предсказание):')
display(shap_df)
```


SHAP показывает не только важность признаков, но и направление их влияния. На диаграмме признаки отсортированы по убыванию важности сверху вниз. Горизонтальная ось - вклад признака в отклонение предсказания от среднего значения: положительный вклад сдвигает предсказание в сторону более высокого ppltrst, отрицательный - в сторону более низкого. Цвет точки кодирует значение признака: красный - высокое, синий - низкое.

Для CPI и eduyrs картина однозначная: чем выше значение признака, тем выше предсказанный уровень доверия у респондента. Для возраста и conservation цвета смешаны по обе стороны нуля - это говорит о том, что эффект этих признаков нелинейный и зависит от конкретных сочетаний значений других переменных.


#### 6.1.5. Бинарная постановка: классификация ppltrst >= 5



Бинарная постановка полезна, когда интерес сосредоточен на отличении "доверяющих" от "недоверяющих" в целом, а не на точном предсказании отметки на шкале. Распределение классов 65/35 - несбалансированное, поэтому мы осторожно работаем с метриками.


```python
# Бинарная классификация: train/test split со стратификацией по целевой переменной
X_tr_c, X_te_c, y_tr_c, y_te_c = train_test_split(
    X, y_bin, test_size=0.2, stratify=y_bin, random_state=RANDOM_STATE
)
cv_c = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
print(f'Train: {len(X_tr_c):,}, Test: {len(X_te_c):,}, доля класса 1 в train: {y_tr_c.mean():.3f}')

# Подбор гиперпараметров для RF-classifier и XGB-classifier
rf_cls_gs = GridSearchCV(
    RandomForestClassifier(n_jobs=-1, random_state=RANDOM_STATE),
    rf_grid, scoring='f1_macro', cv=cv_c, n_jobs=-1, refit=True,
)
with suppress_stderr():
    rf_cls_gs.fit(X_tr_c, y_tr_c)

xgb_cls_gs = GridSearchCV(
    XGBClassifier(random_state=RANDOM_STATE, n_jobs=-1, verbosity=0,
                  tree_method='hist', eval_metric='logloss'),
    xgb_grid, scoring='f1_macro', cv=cv_c, n_jobs=-1, refit=True,
)
with suppress_stderr():
    xgb_cls_gs.fit(X_tr_c, y_tr_c)
print(f'RF classifier best: {rf_cls_gs.best_params_}, CV F1_macro = {rf_cls_gs.best_score_:.4f}')
print(f'XGB classifier best: {xgb_cls_gs.best_params_}, CV F1_macro = {xgb_cls_gs.best_score_:.4f}')
```

```python
# 4 модели-классификатора, метрики на test-выборке
logreg = Pipeline([('sc', StandardScaler()),
                   ('est', LogisticRegression(max_iter=2000, n_jobs=-1))])

models_cls = {
    'DummyClassifier (most_frequent)': DummyClassifier(strategy='most_frequent',
                                                        random_state=RANDOM_STATE),
    'Logistic Regression': logreg,
    'Random Forest (tuned)': rf_cls_gs.best_estimator_,
    'XGBoost (tuned)': xgb_cls_gs.best_estimator_,
}

rows_cls = []
fitted_cls = {}
for name, mdl in models_cls.items():
    with suppress_stderr():
        mdl.fit(X_tr_c, y_tr_c)
    y_pred = mdl.predict(X_te_c)
    row = {
        'model': name,
        'test_accuracy': accuracy_score(y_te_c, y_pred),
        'test_f1_macro': f1_score(y_te_c, y_pred, average='macro'),
        'test_f1_class0': f1_score(y_te_c, y_pred, pos_label=0),
        'test_f1_class1': f1_score(y_te_c, y_pred, pos_label=1),
    }
    if hasattr(mdl, 'predict_proba'):
        try:
            proba = mdl.predict_proba(X_te_c)[:, 1]
            row['test_roc_auc'] = roc_auc_score(y_te_c, proba)
            row['test_brier'] = brier_score_loss(y_te_c, proba)
        except ValueError:
            row['test_roc_auc'] = np.nan
            row['test_brier'] = np.nan
    else:
        row['test_roc_auc'] = np.nan
        row['test_brier'] = np.nan
    rows_cls.append(row)
    fitted_cls[name] = (mdl, y_pred)

cls_results = pd.DataFrame(rows_cls).round(3)
display(cls_results)
```

```python
# Confusion matrices для всех 4 моделей на test-выборке
fig, axes = plt.subplots(1, len(fitted_cls), figsize=(3.5 * len(fitted_cls), 4))
for ax, (name, (_, y_pred)) in zip(axes, fitted_cls.items()):
    cm = confusion_matrix(y_te_c, y_pred)
    ax.imshow(cm, cmap='Blues')
    for i in range(2):
        for j in range(2):
            ax.text(j, i, cm[i, j], ha='center', va='center',
                    color='white' if cm[i, j] > cm.max() / 2 else 'black')
    ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
    ax.set_xticklabels(['pred 0', 'pred 1'])
    ax.set_yticklabels(['true 0', 'true 1'])
    ax.set_title(name, fontsize=10)
plt.tight_layout(); plt.show()
```


F1 по мажоритарному классу (класс 1) у всех моделей около 0.79, в том числе у DummyClassifier. Это следствие дисбаланса классов: стратегия "всегда предсказывать 1" даёт полный recall на мажоритарном классе при разумной precision, что и даёт высокий F1. Метрика выглядит хорошо, но не различает модели со слабым и сильным сигналом.

Метрики, чувствительные к реальному сигналу:

- **ROC-AUC.** Dummy = 0.50 (случайное гадание), LogReg = 0.69, Random Forest = 0.71, XGBoost = 0.72.
- **F1_macro** (среднее F1 по обоим классам). Dummy = 0.40, XGBoost = 0.62.
- **Brier score** (чем ниже, тем лучше калибровка вероятностных предсказаний). XGBoost = 0.196, Random Forest = 0.198, LogReg = 0.205.

По этим метрикам XGBoost немного лучше RF и заметно лучше LogReg, а Dummy остаётся на уровне случайного дискриминатора. Прирост нелинейных моделей над линейной умеренный: классификация по ppltrst на 15 признаках - задача со средним сигналом.


#### 6.1.6. Калибровка вероятностей


```python
# Calibration curve: разбиваем предсказанные вероятности на 10 бинов (по квантилям)
# и смотрим, насколько средняя предсказанная вероятность в бине соответствует реальной доле класса 1.
fig, ax = plt.subplots(figsize=(7, 6))
ax.plot([0, 1], [0, 1], 'k--', label='идеальная калибровка')
colors = {'Logistic Regression': 'steelblue',
          'Random Forest (tuned)': 'darkorange',
          'XGBoost (tuned)': 'crimson'}
brier_rows = []
for name, (mdl, _) in fitted_cls.items():
    if 'Dummy' in name or not hasattr(mdl, 'predict_proba'):
        continue
    proba = mdl.predict_proba(X_te_c)[:, 1]
    frac_pos, mean_pred = calibration_curve(y_te_c, proba, n_bins=10, strategy='quantile')
    brier = brier_score_loss(y_te_c, proba)
    brier_rows.append({'model': name, 'brier_score': brier})
    ax.plot(mean_pred, frac_pos, marker='o', color=colors.get(name, 'gray'),
            label=f'{name} (Brier = {brier:.3f})')
ax.set_xlabel('Средняя предсказанная вероятность')
ax.set_ylabel('Доля класса 1 в бине')
ax.set_title('Калибровка вероятностей, бинарная задача (test set)')
ax.legend()
plt.tight_layout(); plt.show()

display(pd.DataFrame(brier_rows).round(4))
```


Идеально откалиброванная модель должна лежать на диагонали: предсказанная вероятность 0.7 соответствует реальной доле класса 1, равной 0.7 в таком бине. LogReg лежит близко к диагонали - её вероятностные предсказания можно интерпретировать напрямую. RF и XGBoost в зонах высокой вероятности отклоняются вниз: когда модель говорит 0.8, на самом деле доля класса 1 примерно 0.75. Небольшая недооценка уверенности.

Brier score у XGBoost минимальный (0.196), у RF близкий (0.198), у LogReg чуть выше (0.205). Разница между тремя моделями небольшая. Если прикладная задача требует интерпретируемых вероятностей (например, "с какой вероятностью этот респондент попадёт в группу доверяющих"), LogReg выбирается за счёт более точной калибровки без дополнительной процедуры isotonic или Platt.


#### 6.1.7. Подбор порога


```python
# ROC-кривая и оптимальный порог для XGBoost classifier
xgb_clf, _ = fitted_cls['XGBoost (tuned)']
proba = xgb_clf.predict_proba(X_te_c)[:, 1]
fpr, tpr, thr_arr = roc_curve(y_te_c, proba)

# Оптимальный порог: максимум разности tpr - fpr (критерий Юдена / youden's J)
score = tpr - fpr
i_opt = int(score.argmax())
t_opt = float(thr_arr[i_opt])

# Метрики при пороге 0.5 и при оптимальном пороге
y_def = (proba >= 0.5).astype(int)
y_opt = (proba >= t_opt).astype(int)

thr_table = pd.DataFrame([
    {'threshold': 0.5,
     'accuracy': accuracy_score(y_te_c, y_def),
     'f1_macro': f1_score(y_te_c, y_def, average='macro'),
     'f1_class1': f1_score(y_te_c, y_def, pos_label=1),
     'f1_class0': f1_score(y_te_c, y_def, pos_label=0),
     'recall_class0': confusion_matrix(y_te_c, y_def)[0, 0] / (y_te_c == 0).sum()},
    {'threshold': t_opt,
     'accuracy': accuracy_score(y_te_c, y_opt),
     'f1_macro': f1_score(y_te_c, y_opt, average='macro'),
     'f1_class1': f1_score(y_te_c, y_opt, pos_label=1),
     'f1_class0': f1_score(y_te_c, y_opt, pos_label=0),
     'recall_class0': confusion_matrix(y_te_c, y_opt)[0, 0] / (y_te_c == 0).sum()},
]).round(4)
display(thr_table)
print(f'Оптимальный порог (argmax tpr - fpr): {t_opt:.3f}')
```

```python
# ROC-кривая с отмеченным оптимальным порогом
fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(fpr, tpr, color='crimson', lw=1.5, label='ROC')
ax.plot([0, 1], [0, 1], 'k--', lw=0.8, label='случайное гадание')
ax.scatter([fpr[i_opt]], [tpr[i_opt]], color='black', zorder=5, s=60,
           label=f'Оптимальный порог = {t_opt:.2f}')
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC XGBoost classifier (test set)')
ax.legend()
plt.tight_layout(); plt.show()
```


Порог 0.5 - стандартное значение для бинарной классификации, оптимальное при сбалансированных классах. У нас распределение 65/35 в пользу класса 1, и модель минимизирует общую ошибку, охотнее предсказывая класс 1. В результате класс 0 распознаётся плохо: F1 класса 0 = 0.45, recall класса 0 = 0.35 (модель находит только треть "недоверяющих"), а F1 класса 1 высокий за счёт большого количества правильных предсказаний мажоритарного класса.

Чтобы модель одинаково хорошо различала оба класса, порог нужно поднять: тогда для предсказания "1" требуется более высокая оценка вероятности, и часть наблюдений с низкой уверенностью переходит в класс 0. Оптимальный порог подбираем по критерию "максимум разности tpr - fpr" на ROC-кривой, где tpr (true positive rate) - доля правильно найденных представителей класса 1 среди всех респондентов с реальным классом 1 (чувствительность), а fpr (false positive rate) - доля респондентов с реальным классом 0, ошибочно отнесённых к классу 1. Для нашего XGBoost оптимальный порог равен 0.60.

При этом пороге F1_macro поднимается с 0.62 до 0.66, recall класса 0 - с 0.35 до 0.58 (модель находит больше половины "недоверяющих"), F1 класса 1 немного снижается (0.79 -> 0.76). Общая accuracy проседает с 0.70 до 0.69 - это плата за сбалансированность.

Какой порог выбрать - зависит от стоимости ошибок. Если одинаково важно различать оба класса, порог 0.60 предпочтителен. Если цель - общая точность предсказаний, 0.5 лучше. В таблице приводим оба варианта.



### 6.2. Многоуровневый подход

Многоуровневая регрессия оценивается через класс `statsmodels.MixedLM` (mixed linear model, смешанная линейная модель) методом максимального правдоподобия. Мы работаем на 29 странах - это нижняя граница диапазона, в котором частотная оценка параметров на уровне 2 (страновом) даёт несмещённые результаты. Страновые коэффициенты имеют повышенную погрешность по сравнению с индивидуальными, поэтому интерпретируем их консервативно.


#### 6.2.1. Null model: ICC


```python
import statsmodels.formula.api as smf

def extract_coefs(res):
    """Таблица коэффициентов MixedLM: coef, se, z, p, sig."""
    df_out = pd.DataFrame({
        'coef': res.params, 'se': res.bse, 'z': res.tvalues, 'p': res.pvalues,
    }).round(4)
    return df_out

def format_sig(p):
    if pd.isna(p): return ''
    if p < 0.001: return '***'
    if p < 0.01: return '**'
    if p < 0.05: return '*'
    if p < 0.10: return '.'
    return ''

# M0: нулевая модель, только страновой случайный свободный член
m0_model = smf.mixedlm('ppltrst ~ 1', sample, groups=sample['cntry'])
m0_res = m0_model.fit(reml=False)

null_tau = float(m0_res.cov_re.iloc[0, 0])
null_sigma = float(m0_res.scale)
icc = null_tau / (null_tau + null_sigma)

m0 = extract_coefs(m0_res).reset_index().rename(columns={'index': 'term'})
display(m0)
print(f'tau0² (между-страновая дисперсия): {null_tau:.4f}')
print(f'sigma² (внутри-страновая дисперсия): {null_sigma:.4f}')
print(f'ICC = tau0² / (tau0² + sigma²) = {icc:.4f}')
print(f'LogLik = {m0_res.llf:.2f}, AIC = {m0_res.aic:.2f}, BIC = {m0_res.bic:.2f}')
```


Null-модель не содержит предикторов, только страновой random intercept. Она раскладывает дисперсию ppltrst на два уровня. ICC = 0.171 показывает, что 17% общей дисперсии приходится на различия между странами, 83% - на индивидуальные различия внутри страны.

Значение ICC выше 0.05 обосновывает multilevel-подход (Hox 2018). **H3 подтверждается.**


#### 6.2.2. Индивидуальные предикторы (M1)


```python
# M1: индивидуальные предикторы (ценности, медиа, демография) + страновой random intercept
m1_formula = ('ppltrst ~ openness_z + conservation_z + self_trans_z + self_enh_z'
              ' + netusoft_z + log_netustm_z + log_nwspol_z'
              ' + agea_z + female + eduyrs_z + hinctnta_z + domicil_z')
m1_res = smf.mixedlm(m1_formula, sample, groups=sample['cntry']).fit(reml=False)
m1_tau = float(m1_res.cov_re.iloc[0, 0])
m1_sigma = float(m1_res.scale)

m1 = extract_coefs(m1_res).reset_index().rename(columns={'index': 'term'})
m1['sig'] = m1['p'].apply(format_sig)
display(m1)
print(f'tau0² = {m1_tau:.4f}, sigma² = {m1_sigma:.4f}')
print(f'LogLik = {m1_res.llf:.2f}, AIC = {m1_res.aic:.2f}')
```

```python
# Forest plot коэффициентов M1 с 95% доверительными интервалами
m1_fixed = m1[~m1['term'].isin(['Intercept', 'Group Var'])].sort_values('coef')

fig, ax = plt.subplots(figsize=(9, 6))
y = range(len(m1_fixed))
ax.errorbar(m1_fixed['coef'], y, xerr=1.96 * m1_fixed['se'],
            fmt='o', color='steelblue', capsize=3)
ax.axvline(0, color='gray', linestyle='--', alpha=0.6)
ax.set_yticks(list(y))
ax.set_yticklabels(m1_fixed['term'])
ax.set_xlabel('Коэффициент (95% CI), стандартизованная шкала')
ax.set_title('M1: индивидуальные предикторы социального доверия')
plt.tight_layout(); plt.show()
```


**Ценности.** Self-transcendence положительно связан с доверием (+0.23, p < 0.001): более сильная ориентация на заботу о других и равенство соответствует более высокому генерализованному доверию. Conservation значимо отрицательно (-0.26, p < 0.001): ориентация на стабильность и близкий круг снижает доверие к большинству. Self-enhancement и openness значимой связи не показали. Знаки self-transcendence и conservation согласуются с Uslaner (2002).

**Медиа.** netusoft положителен (+0.16, p < 0.001), log_nwspol положителен (+0.09, p < 0.001). log_netustm имеет маргинальный отрицательный эффект (p = 0.08). Контент важнее количества: сам факт онлайн-активности и информированность о политике связаны с доверием положительно, а длительное неструктурированное потребление - нет.

**Демография.** Образование даёт самый большой индивидуальный эффект (+0.22 SD, p < 0.001), затем доход (+0.17) и возраст (+0.12). Пол и тип населённого пункта при контроле на остальные переменные значимо не влияют.


#### 6.2.3. Страновые предикторы (M2)


```python
# M2: M1 + три страновых предиктора (GDP per capita PPP, Gini, CPI)
m2_formula = m1_formula + ' + gdp_pcap_ppp_z + gini_z + cpi_z'
m2_res = smf.mixedlm(m2_formula, sample, groups=sample['cntry']).fit(reml=False)
m2_tau = float(m2_res.cov_re.iloc[0, 0])
m2_sigma = float(m2_res.scale)

m2 = extract_coefs(m2_res).reset_index().rename(columns={'index': 'term'})
m2['sig'] = m2['p'].apply(format_sig)
display(m2)
print(f'tau0² = {m2_tau:.4f}, sigma² = {m2_sigma:.4f}')
print(f'LogLik = {m2_res.llf:.2f}, AIC = {m2_res.aic:.2f}')
```

```python
# Forest plot M2 с цветовой разметкой: синий - индивидуальный уровень, красный - страновой
m2_fixed = m2[~m2['term'].isin(['Intercept', 'Group Var'])].sort_values('coef')

fig, ax = plt.subplots(figsize=(9, 6))
y = range(len(m2_fixed))
is_country = m2_fixed['term'].isin(['gdp_pcap_ppp_z', 'gini_z', 'cpi_z'])
for yi, ci, se, c in zip(y, m2_fixed['coef'], m2_fixed['se'], is_country):
    col = 'crimson' if c else 'steelblue'
    ax.errorbar(ci, yi, xerr=1.96 * se, fmt='o', color=col, capsize=3)
ax.axvline(0, color='gray', linestyle='--', alpha=0.6)
ax.set_yticks(list(y))
ax.set_yticklabels(m2_fixed['term'])
ax.set_xlabel('Коэффициент (95% CI), стандартизованная шкала')
ax.set_title('M2: полная модель (синий - индивидуальный уровень, красный - страновой)')
plt.tight_layout(); plt.show()
```


CPI - самый сильный страновый предиктор (+0.54, p < 0.001). Каждое стандартное отклонение улучшения CPI соответствует 0.54 пунктам доверия на шкале 0-10 у индивида.

GDP per capita (+0.17) и Gini (-0.10) имеют ожидаемые знаки, но после контроля на CPI их коэффициенты не достигают значимости. Это согласуется с институциональным подходом (Rothstein, Uslaner 2005) и с корреляцией CPI-GDP из раздела 3.4 (r = 0.81): экономический уровень и институциональное качество на уровне 29 стран почти коллинеарны, и общая вариация объясняется преимущественно через CPI.

Индивидуальные коэффициенты при добавлении страновых переменных почти не изменились - это говорит об устойчивости выводов раздела 6.2.2.


#### 6.2.4. Декомпозиция дисперсии


```python
# Сколько между-страновой и внутри-страновой дисперсии объясняют модели M1 и M2
# Pseudo-R² по Snijders-Bosker: (tau_null - tau_model) / tau_null
var_df = pd.DataFrame([
    ['Between-country variance (tau0²)', null_tau, m1_tau, m2_tau],
    ['Within-country variance (sigma²)', null_sigma, m1_sigma, m2_sigma],
], columns=['component', 'M0_null', 'M1_individual', 'M2_full']).round(4)
display(var_df)

pct_between_m1 = (null_tau - m1_tau) / null_tau * 100
pct_between_m2 = (null_tau - m2_tau) / null_tau * 100
pct_within_m1 = (null_sigma - m1_sigma) / null_sigma * 100

print(f'M1 объяснено межстрановой дисперсии: {pct_between_m1:.1f}%')
print(f'M1 объяснено внутри-страновой дисперсии: {pct_within_m1:.1f}%')
print(f'M2 объяснено межстрановой дисперсии: {pct_between_m2:.1f}%')
```

```python
# Визуализация: сокращение tau0² при переходе M0 -> M1 -> M2
fig, ax = plt.subplots(figsize=(7, 4))
model_names = ['M0 (null)', 'M1 (individual)', 'M2 (+ country)']
vals = [null_tau, m1_tau, m2_tau]
ax.bar(model_names, vals, color=['steelblue', 'cornflowerblue', 'crimson'])
ax.set_ylabel('Между-страновая дисперсия (tau0²)')
ax.set_title('Сокращение между-страновой дисперсии по моделям')
for i, v in enumerate(vals):
    ax.text(i, v + 0.02, f'{v:.3f}', ha='center')
plt.tight_layout(); plt.show()

# Информационные критерии всех трёх моделей
ic = pd.DataFrame([
    {'model': 'M0_null', 'logLik': m0_res.llf, 'AIC': m0_res.aic, 'BIC': m0_res.bic,
     'n_params': len(m0_res.params)},
    {'model': 'M1_individual', 'logLik': m1_res.llf, 'AIC': m1_res.aic, 'BIC': m1_res.bic,
     'n_params': len(m1_res.params)},
    {'model': 'M2_full', 'logLik': m2_res.llf, 'AIC': m2_res.aic, 'BIC': m2_res.bic,
     'n_params': len(m2_res.params)},
]).round(2)
ic['ICC_null'] = round(icc, 4)
print('Информационные критерии:')
display(ic)
```


Три страновых переменных объясняют 79% межстрановой дисперсии доверия; основной вклад - CPI. Страновые различия в доверии почти полностью объясняются качеством институтов (с контролем на индивидуальные характеристики). Внутри-страновая дисперсия снижается всего на 4% при добавлении индивидуальных предикторов - типичный результат для моделей доверия: большая часть индивидуальной вариации остаётся необъяснённой и приходится на личный опыт и психологические факторы, не покрытые ESS.

AIC и BIC последовательно снижаются от M0 к M2.


#### 6.2.5. Диагностики финальной модели


```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# VIF считаем на двух уровнях:
# Level 1: индивидуальные предикторы (N = 44200) - проверяем коллинеарность внутри страны
# Level 2: три страновых переменных (N = 29) - проверяем коллинеарность GDP/Gini/CPI

l1_cols = [
    'openness_z', 'conservation_z', 'self_trans_z', 'self_enh_z',
    'netusoft_z', 'log_netustm_z', 'log_nwspol_z',
    'agea_z', 'female', 'eduyrs_z', 'hinctnta_z', 'domicil_z',
]
X1 = sample[l1_cols].to_numpy()
vif_l1 = pd.DataFrame({
    'predictor': l1_cols,
    'VIF': [variance_inflation_factor(X1, i) for i in range(X1.shape[1])],
    'level': 'individual',
})

country_unique = sample[['cntry', 'gdp_pcap_ppp_z', 'gini_z', 'cpi_z']].drop_duplicates('cntry')
X2 = country_unique[['gdp_pcap_ppp_z', 'gini_z', 'cpi_z']].to_numpy()
vif_l2 = pd.DataFrame({
    'predictor': ['gdp_pcap_ppp_z', 'gini_z', 'cpi_z'],
    'VIF': [variance_inflation_factor(X2, i) for i in range(X2.shape[1])],
    'level': 'country',
})

vif = pd.concat([vif_l1, vif_l2], ignore_index=True).round(3)
print('VIF предикторов M2:')
display(vif)

print()
print(f'Максимальный VIF (индивидуальный уровень): {vif[vif["level"] == "individual"]["VIF"].max():.3f}')
print(f'Максимальный VIF (страновой уровень): {vif[vif["level"] == "country"]["VIF"].max():.3f}')
```


На индивидуальном уровне максимальные VIF (openness 5.0 и self_trans 5.2) лежат на границе порогового значения 5.0. Причина в том, что четыре композитных ценностных индекса построены из пересекающихся факторов Шварца (например, openness и self_trans используют часть одних и тех же items через универсалистские нагрузки). Это не критично: коэффициенты интерпретируются корректно, а направления знаков совпадают с теорией.

На уровне стран VIF составляет от 1.2 (Gini) до 3.0 (CPI) и 2.9 (GDP). Коллинеарность присутствует, но не чрезмерна. Оценки коэффициентов достаточно стабильны при том ограничении N = 29 стран.

### 6.3. Сопоставление подходов

Собираем для каждого из 15 признаков два показателя важности: абсолютное значение стандартизованного коэффициента в модели M2 и permutation importance на test-выборке (отдельно для Random Forest и XGBoost). Если два независимых подхода дают схожий порядок признаков по важности, это свидетельствует в пользу устойчивости содержательных выводов. Согласованность рангов оцениваем коэффициентом Spearman ρ.


```python
# Собираем полную таблицу сопоставления: multilevel M2 коэффициенты + ML permutation importance
TERM_TO_FEATURE = {
    'openness_z': 'openness', 'conservation_z': 'conservation',
    'self_trans_z': 'self_trans', 'self_enh_z': 'self_enh',
    'netusoft_z': 'netusoft', 'log_netustm_z': 'log_netustm',
    'log_nwspol_z': 'log_nwspol', 'agea_z': 'agea',
    'female': 'female', 'eduyrs_z': 'eduyrs',
    'hinctnta_z': 'hinctnta', 'domicil_z': 'domicil',
    'gdp_pcap_ppp_z': 'gdp_pcap_ppp', 'gini_z': 'gini', 'cpi_z': 'cpi',
}

mlm = m2.copy()
mlm['feature'] = mlm['term'].map(TERM_TO_FEATURE)
mlm = mlm.dropna(subset=['feature']).copy()
mlm['abs_coef'] = mlm['coef'].abs()
mlm['sig_flag'] = mlm['p'] < 0.05

rf_ic = rf_imp.rename(columns={'importance_mean': 'importance_RF'})[['feature', 'importance_RF']]
xgb_ic = xgb_imp.rename(columns={'importance_mean': 'importance_XGB'})[['feature', 'importance_XGB']]

full = (mlm[['feature', 'coef', 'abs_coef', 'p', 'sig_flag']]
        .merge(rf_ic, on='feature')
        .merge(xgb_ic, on='feature'))
full['rank_coef'] = full['abs_coef'].rank(ascending=False).astype(int)
full['rank_RF'] = full['importance_RF'].rank(ascending=False).astype(int)
full['rank_XGB'] = full['importance_XGB'].rank(ascending=False).astype(int)
full = full.sort_values('abs_coef', ascending=False).reset_index(drop=True)

cols_show = ['feature', 'coef', 'p', 'sig_flag',
             'importance_RF', 'importance_XGB',
             'rank_coef', 'rank_RF', 'rank_XGB']
display(full[cols_show].round(4))
```

```python
# Spearman ρ между рангом |coef_M2| и permutation importance (RF и XGBoost)
from scipy.stats import spearmanr

rho_rf, p_rf = spearmanr(full['abs_coef'].to_numpy(), full['importance_RF'].to_numpy())
rho_xgb, p_xgb = spearmanr(full['abs_coef'].to_numpy(), full['importance_XGB'].to_numpy())

spearman_df = pd.DataFrame([
    {'comparison': '|coef_M2| vs importance_RF', 'spearman_rho': rho_rf, 'p_value': p_rf},
    {'comparison': '|coef_M2| vs importance_XGB', 'spearman_rho': rho_xgb, 'p_value': p_xgb},
]).round(4)
display(spearman_df)
```

```python
# Scatterplot |coef_M2| vs permutation importance (RF). Цвет: красный = значимо в M2 (p < 0.05)
fig, ax = plt.subplots(figsize=(9, 6))
colors = ['crimson' if s else 'lightgray' for s in full['sig_flag']]
ax.scatter(full['abs_coef'], full['importance_RF'], c=colors, s=70)
for _, r in full.iterrows():
    ax.annotate(r['feature'], (r['abs_coef'], r['importance_RF']),
                xytext=(5, 5), textcoords='offset points', fontsize=9)
ax.set_xlabel('|Коэффициент multilevel M2|, стандартизованная шкала')
ax.set_ylabel('Permutation importance Random Forest')
ax.set_title('Сопоставление рангов признаков в двух подходах\n(красный - значим в M2 при p < 0.05)')
plt.tight_layout(); plt.show()
```


Spearman ρ между |coef_M2| и permutation importance Random Forest равен 0.75 (p = 0.0014), с XGBoost - 0.80 (p = 0.0003). Оба значения значимы на p < 0.01: порядок признаков по важности в двух методах в основном совпадает.

**Совпадение в верхней части рейтинга.** CPI занимает первое место и в многоуровневой модели, и в обеих ML-моделях. Образование, доход домохозяйства, частота использования интернета и возраст стабильно находятся в верхней половине рейтинга у обоих методов. Основной содержательный вывод (доминирование CPI среди страновых предикторов и значимый вклад ресурсных характеристик: образование, доход) не зависит от выбора методологии.

**Два расхождения, которые нужно объяснить.**

Ценности conservation и self-transcendence в многоуровневой модели занимают 2-е и 3-е места по абсолютной величине коэффициента, но в permutation importance Random Forest опускаются на 8-9 позиции. Причина техническая: линейный эффект ценностей хорошо передаётся коэффициентом регрессии, но в tree-based моделях при перестановке признака R² падает на меньшую величину, поскольку основной вклад в предсказание приходится на страновые признаки (CPI, GDP).

Для GDP и Gini ситуация обратная: в permutation importance они поднимаются на 2-4 места, а в многоуровневой модели после контроля на CPI теряют значимость. Это следствие нелинейных эффектов, которые RF и XGBoost способны выделить (например, различный вклад GDP в странах с низким и высоким уровнем дохода), а линейная многоуровневая модель объединяет с сигналом CPI из-за их высокой корреляции (r = 0.81).



### 6.4. Итоги Части 2

- **H3 подтверждается.** ICC = 0.171: страновой уровень объясняет значимую, но не доминирующую долю дисперсии доверия.
- **H4 подтверждается частично.** Self-transcendence положительно, conservation отрицательно связаны с доверием, что соответствует теории. Для self-enhancement и openness значимой связи не обнаружено. Из четырёх высших измерений Шварца проверку проходят два.
- **H5 подтверждается частично.** Среди страновых предикторов доминирует CPI (+0.54, p < 0.001). GDP и Gini после контроля на CPI теряют значимость: их эффект опосредуется институциональным качеством.
- **Декомпозиция дисперсии.** Индивидуальные предикторы объясняют 28% межстрановой и 4% внутри-страновой вариации; страновые предикторы поднимают объяснение межстрановой вариации до 79%.
- **Диагностики финальной модели.** VIF в пределах допустимого на обоих уровнях, остатки близки к нормальному распределению.
- **Сопоставление ML и многоуровневой регрессии.** Spearman ρ между рангами |coef| и permutation importance равен 0.75 (Random Forest) и 0.81 (XGBoost), оба значения значимы на p < 0.01. Ведущие предикторы (CPI, eduyrs, hinctnta, netusoft, agea) совпадают в обоих подходах. Расхождения в ранжировании ценностей и стран по GDP/Gini связаны с тем, как два метода обрабатывают линейные и нелинейные эффекты.

### 7.2. Содержательная интерпретация

Результаты согласованно указывают на институциональную природу генерализованного доверия в Европе. Из трёх классов объяснительных переменных - ценности, индивидуальные ресурсы, страновые характеристики - именно CPI даёт самый сильный и устойчивый эффект. Три страновых предиктора вместе объясняют 79% межстрановой дисперсии; внутри страны большая часть вариации остаётся необъяснённой.

На индивидуальном уровне работают классические предикторы. Образование, доход и возраст связаны с более высоким доверием положительно. Это согласуется с результатами Delhey, Newton (2005) и Uslaner (2002). Ценности вносят самостоятельный вклад сверх ресурсных переменных: альтруистическая ориентация (self-transcendence) усиливает доверие, ориентация на стабильность и близкий круг (conservation) ослабляет.

По медиапотреблению картина двойственная. Частота использования интернета и время чтения политических новостей (log_nwspol) положительно связаны с доверием, а общее время в интернете - слабо отрицательно. Это соответствует идее Delhey и Newton (2005) об информационных эффектах: важнее качество потребляемой информации, чем её общий объём.

Сопоставление двух подходов на одних и тех же данных даёт Spearman ρ = 0.75-0.81. Ведущие предикторы (CPI, eduyrs, hinctnta, netusoft) совпадают и в многоуровневой регрессии, и в ML-моделях. Значит, содержательные выводы не зависят от выбора методологии. Расхождения по ценностям и по GDP/Gini объясняются различной чувствительностью двух подходов: многоуровневая модель хорошо передаёт линейные эффекты ценностных признаков через коэффициенты регрессии, а Random Forest и XGBoost учитывают нелинейные эффекты GDP и Gini, которые линейная модель не различает из-за их высокой корреляции с CPI.



### 7.3. Ограничения

Разделены на три категории.

**Методологические:**

- **Ordinal-структура PVQ** (порядковая шкала). Итемы имеют 6 категорий, формально это порядковые переменные. В работе они трактуются как континуальные (непрерывные) при оценке CFA - стандартное приближение для 5 и более категорий при приблизительно симметричных распределениях. Более строгий подход через WLSMV (weighted least squares mean and variance adjusted, взвешенный метод наименьших квадратов со скорректированными средним и дисперсией) требует перехода на R/lavaan и в данной работе не применяется.
- **Скалярная инвариантность формально не тестировалась.** Пакет semopy 2.3.11 не поддерживает constrained multigroup CFA (multigroup CFA с ограничениями на равенство параметров между группами). Для целей работы (регрессионные коэффициенты ценностей на доверие) достаточно достигнутой частичной метрической инвариантности; прямое сравнение абсолютных средних значений ценностей между странами не проводится.
- **N = 29 стран на уровне 2.** 29 стран - нижняя граница рекомендованного диапазона для частотной оценки параметров. Страновые коэффициенты имеют повышенную погрешность.
- **Модель с random intercept.** Использована модель со случайным свободным членом по стране. Вопрос о вариации эффектов ценностей между странами (random slopes) выходит за рамки работы.

**Данные:**

- **Кросс-секционный дизайн.** Один срез ESS R9 (2018). Оцениваются связи, но не причинно-следственные отношения: положительный коэффициент CPI не означает, что улучшение институтов автоматически повысит доверие - возможна и обратная каузальность.
- **Sampling weights** (pspwght, pweight) не использованы. Для оценки регрессионных параметров это стандартное упрощение в прикладных работах; для оценок популяционных статистик - ограничение.
- **Listwise deletion** по 12 ключевым переменным оставляет 44200 из 49519 респондентов (потеря 10.7%). Пропуски распределены по странам относительно равномерно, сильного смещения не ожидается.
- **Медианная замена пропусков в hinctnta и netustm.** 19.5% и 30.4% пропусков заменены медианой внутри страны. Это снижает вариацию соответствующих переменных и может недооценивать их связь с доверием.

**Обобщаемость:**

- **Только Европа.** 29 стран ESS дают ограниченный культурный контекст. На страны за пределами Европы модель не переносится автоматически.
- **Ограниченный набор страновых переменных.** Использованы только GDP, Gini, CPI. Альтернативные наборы (V-Dem, Freedom House, Hofstede) могут добавить измерения, связанные с демократией и культурой.
- **Обобщение на новую страну.** Проверка переноса модели на hold-out страну (leave-one-country-out валидация) в работе не проводилась и является направлением дальнейших исследований.



### 7.4. Вклад работы

1. **Последовательное применение трёх методов на одних данных.** Работа показывает, как CFA с тестом инвариантности, многоуровневая регрессия и ML-модели с SHAP могут быть встроены в единый pipeline: факторные оценки из CFA используются как признаки и в multilevel-регрессии, и в ML-моделях. В большинстве прикладных работ по социальному доверию применяется только одна из трёх методологий.
2. **Численное сопоставление ML и multilevel.** Spearman ρ между рангом |coef_M2| и permutation importance даёт количественный ответ на вопрос, насколько выводы устойчивы к выбору подхода. На ESS R9 ρ = 0.75-0.81 (p < 0.01).
3. **Выявление CPI как доминирующего странового предиктора доверия.** После контроля на CPI экономический уровень (GDP) и неравенство (Gini) не дают самостоятельного вклада. Это согласуется с институциональной перспективой в литературе и количественно формализуется на полном наборе стран ESS.



### 7.5. Направления дальнейших исследований

- **Случайные наклоны (random slopes) для ценностей.** Проверка того, различаются ли эффекты self-transcendence и conservation между странами сверх случайной ошибки.
- **Leave-one-country-out валидация** (кросс-валидация с поочерёдным исключением одной страны из обучающей выборки). Оценка того, насколько модель переносится на новую страну, когда она обучается на 28 остальных.
- **Оценка через REML** (restricted maximum likelihood, ограниченный метод максимального правдоподобия) для менее смещённой оценки дисперсионных компонент по сравнению с обычным методом максимального правдоподобия.
- **Формальный тест скалярной инвариантности через R/lavaan** с WLSMV-оценкой для порядковых PVQ-итемов.
- **Межуровневые взаимодействия (cross-level interactions).** Например, усиливается ли связь self-transcendence с доверием в странах с высокой коррупцией.
- **Множественная подстановка пропусков (multiple imputation)** через mice или miceforest как анализ чувствительности к listwise deletion.
- **Весовые коэффициенты выборки (sampling weights: pspwght)** через взвешенный метод псевдоправдоподобия (weighted pseudo-likelihood).
- **Дополнительные страновые переменные** (V-Dem, Hofstede IDV) для расширения объяснительной мощности модели.



## 8. Выводы

В работе проведён кросс-национальный анализ структуры базовых ценностей Шварца и их связи с генерализованным социальным доверием в 29 странах Европы на данных European Social Survey Round 9 (2018). Анализ разделён на две содержательные части с разными методологиями: структурный анализ ценностей через CFA и тест инвариантности (Часть 1), затем моделирование доверия двумя взаимодополняющими подходами (многоуровневой регрессией и машинным обучением) с последующим сопоставлением результатов (Часть 2).

**Структура ценностей.** Десятифакторная первичная модель Шварца воспроизводится на объединённой европейской выборке (CFI = 0.906, RMSEA = 0.058, SRMR = 0.050). Конфигуральная инвариантность подтверждается в 28 из 29 стран с CFI >= 0.85; единственная страна с маргинальным значением - Словения (CFI = 0.840). По частичной метрической инвариантности 9 из 11 свободно оцениваемых факторных нагрузок устойчивы между странами; два итема (`ipbhprp` в ценности Conformity и `iprspot` в ценности Power) демонстрируют значительную культурную специфику. Формальный тест скалярной инвариантности требует специализированного инструментария и в работе не проводился.

**Социальное доверие.** ICC = 0.171: страновой уровень объясняет 17% общей дисперсии индивидуальных оценок доверия. Из индивидуальных предикторов значимы self-transcendence (+), conservation (-), образование, доход, возраст, частота использования интернета и время чтения политических новостей. Из страновых доминирует индекс восприятия коррупции (CPI): каждое стандартное отклонение улучшения качества институтов соответствует +0.54 пунктам индивидуального доверия. GDP per capita и Gini после контроля на CPI самостоятельной значимой связи не дают: их эффект опосредуется институциональным качеством. Три страновых предиктора вместе объясняют 79% межстрановой вариации доверия.

**Методологический итог.** ML-модели (Random Forest и XGBoost с подбором гиперпараметров через GridSearchCV, permutation importance и SHAP) выделяют ту же иерархию ведущих предикторов, что и многоуровневая регрессия. Spearman ρ между |coef_M2| и permutation importance XGBoost равен 0.81 (p = 0.0003). Расхождения в порядке ценностей и GDP/Gini связаны с тем, что многоуровневая модель лучше передаёт линейные эффекты ценностных признаков, а ML-модели учитывают нелинейные эффекты GDP и Gini, которые линейная модель не различает из-за их высокой корреляции с CPI.

**Содержательный итог.** Генерализованное социальное доверие в Европе в существенной мере определяется качеством институтов страны; внутри каждой страны его формируют ресурсные характеристики (образование, доход) и ценностные ориентации (альтруизм усиливает доверие, ориентация на стабильность и близкий круг - ослабляет). Экономический уровень и неравенство играют роль, но их влияние опосредуется институциональным качеством. Результат согласуется с институциональной перспективой в литературе о социальном доверии (Rothstein and Uslaner 2005; Delhey and Newton 2005; Uslaner 2002).
